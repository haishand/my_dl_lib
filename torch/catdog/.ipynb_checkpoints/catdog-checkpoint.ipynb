```python
import torch   
import os
import cv2
```

```python
from torch.utils.data import Dataset
```

```python
class YOLODataSet(Dataset):
    def __init__(self, image_path, label_path, transform=None):
        self.image_path = image_path
        self.label_path = label_path
        self.image_files = os.listdir(image_path)
        self.transform = transform        
        
    def __getitem__(self, idx):
        file = self.image_files[idx]
        img_file = os.path.join(self.image_path, file)
        label_file = os.path.join(self.label_path, file.replace("jpg", "txt"))
        
        image = cv2.imread(img_file)
        image = cv2.resize(image, (64, 64))
        image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
        image = torch.tensor(image, dtype=torch.float32).permute(2, 0, 1)

        """
        boxes = []
        with open(label_file, "r") as f:
            lines = f.readlines()
            for line in lines:
                class_id, x, y, w, h = map(float, line.strip().split())
                boxes.append([class_id, x, y, w, h])

        boxes = torch.tensor(boxes, dtype=torch.float32)
        """
        boxes = 0
        with open(label_file, "r") as f:
            line = f.readline().strip()
            cls_id, x, y, w, h = map(float, line.split())
            boxes = cls_id

        boxes = torch.tensor([boxes], dtype=torch.float32)
        if self.transform:
            image = self.transform(image)
            
        return image, boxes

    def __len__(self):
        return len(self.image_files)
            
```

```python
ds = YOLODataSet("data/images", "data/labels")
"""
for i in range(len(ds)):
    img, boxes = ds[i]
    print(img.shape, boxes.shape)
"""
```

```python
import torch.nn as nn
import torch.nn.functional as F

class CatDogModel(nn.Module):
    def __init__(self):
        super(CatDogModel, self).__init__()
        self.padding = nn.ZeroPad2d(3)
        self.conv1 = nn.Conv2d(in_channels=3, out_channels=32, kernel_size=3, padding=1)
        self.bn1 = nn.BatchNorm2d(num_features=32)
        self.maxpool = nn.MaxPool2d(2)      
        self.fc1 = nn.LazyLinear(1)
        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        x = self.padding(x)
        x = self.conv1(x)
        x = self.bn1(x)
        x = F.relu(x)
        x = self.maxpool(x)
        x = x.view(x.size(0), -1)
        x = self.fc1(x)
        x = self.sigmoid(x) 
        return x
```

```python
model = CatDogModel()
```

```python
from torch.utils.data import DataLoader

dataloader = DataLoader(ds, batch_size=4, shuffle=True)
criterion = nn.BCEWithLogitsLoss()
opt = torch.optim.Adam(model.parameters(), lr=0.001)
```

```python
nepoch = 10
for ep in range(nepoch):
    model.train()
    running_loss = 0
    for x, y in dataloader:
        opt.zero_grad()
        yhat = model(x)
        loss = criterion(yhat, y)
        running_loss += loss.item()
        loss.backward()
        opt.step()
    print(f"epoch: {ep+1}/{nepoch}, loss: {running_loss/len(dataloader)}")
```

```python
torch.save(model.state_dict(), "model.pth")
```

```python
m = CatDogModel()
m.load_state_dict(torch.load("model.pth"))
m.eval()
```

```python
test_image = cv2.imread("./data/images/cat.2.jpg")
test_image = cv2.resize(test_image, (64, 64))
test_image = cv2.cvtColor(test_image, cv2.COLOR_BGR2RGB)

print(test_image.shape)

test_image_tensor = torch.tensor(test_image, dtype=torch.float32).permute(2, 0, 1)
test_image_tensor = test_image_tensor.unsqueeze(0)
```
